## Honest interpretation (per project Tier rules)

- The panel **separates tumor from non-neoplastic thyroid** with very large effect (Cohen d = −1.77; AUC = 0.96) — the silencing direction is correct and sanity passes.
- Binary RAI-avid vs RAI-refractory test is **directionally consistent** (refractory has lower panel z; AUC = 0.61) but **statistically inconclusive** due to severe label imbalance (avid n = 4 vs refractory n = 35). This is the Tier-2 anchor we have; we are not over-claiming.
- Driver-class Kruskal is non-significant within this small cohort.

Within the tumor compartment, panel scores were directionally lower in RAI-refractory than in RAI-avid samples (medians −0.17 vs +0.10) but the binary comparison was underpowered (avid n = 4; AUC = 0.61)."*

# Robustness audit — bootstrap + LOO + multivariate Cox

## What and why

The headline TERT 4-group survival result (univariate logrank p = 4.92×10⁻⁶) is statistically strong, but a careful biostatistician will immediately ask:

1. Is the signal driven by a few outlier patients?
2. Does it survive **stage and age adjustment** (TERT⁺ patients are older and more advanced)?
3. Is the bootstrap distribution of p-values stable?

The v17_ACTUAL sprint added all three robustness checks. **The outcome is the honest reframe in §R8 of the manuscript**:

> TERT⁺ univariate Cox HR = 6.31 (95% CI 2.34–17.04, p = 3×10⁻⁴) → **multivariate (stage + age + sex) HR = 1.88 (95% CI 0.58–6.09, p = 0.29)**.
>
> 1,000-iteration stratified bootstrap median p = 2.6×10⁻⁵ (95% CI 3.2×10⁻¹⁵ – 0.21; 93% of iterations p < 0.05).
>
> Leave-one-out across all 36 TERT⁺ patients: worst-case p = 7.5×10⁻⁴ (all 36 LOO iterations p < 10⁻³).
